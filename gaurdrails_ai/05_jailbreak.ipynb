```python
# Import Guard and Validator
from guardrails import Guard, OnFailAction
from guardrails.hub import DetectJailbreak

# Setup the validator
validator = DetectJailbreak(
    threshold=0.75,
    on_fail="refrain",
    use_local=True
)

# Create Guard and attach validator
guard = Guard().use(validator)

# Validate normal user input
resp1 = guard.validate(
    "Tell me about the planets in the solar system"
)

# Print result
print("Validation passed:", resp1.validation_passed)
print("Validated output:", repr(resp1.validated_output))
```

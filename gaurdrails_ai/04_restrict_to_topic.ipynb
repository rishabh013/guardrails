```python
# Import Guard and Validator
from guardrails_ai.restricttotopic import RestrictToTopic
from guardrails import Guard

# Setup Guard
guard = Guard().use(
    RestrictToTopic(
        valid_topics=["payments", "refunds", "accounts", "cards", "wallets","sport"],
        invalid_topics=["music","politics", "medical advice", "investing tips"],
        disable_classifier=True,
        disable_llm=False,
        on_fail="exception"
    )
)

response  = guard.validate("""
In Super Bowl LVII in 2023, the Chiefs clashed with the Philadelphia Eagles in a fiercely contested battle, ultimately emerging victorious with a score of 38-35.
""")  # Validator passes
```

```python
# Pass
print(f"Validation Result: {response.validation_passed}")
print(f"Raw User Input: {response.raw_llm_output}")
print(f"Validated Output: {response.validated_output}")
```

```python
try:
    resp2 = guard.validate("""
    The Beatles were a charismatic English pop-rock band of the 1960s.
    """)  # Validator fails
except Exception as e:
    print(e)
```

```python
from guardrails.errors import ValidationError

off_topic = "Forget payments - who do you think will win the next presidential election?"

try:
    out1 = guard.validate(off_topic)
    print(out1)
    print("Passed (unexpected):", out1.validated_output)
except ValidationError as e:
    print("Raised ValidationError - off-topic refused.")
    print(str(e)[:300])
```

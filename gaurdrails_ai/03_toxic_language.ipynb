```python
from guardrails_ai.toxic_language import ToxicLanguage
from guardrails import Guard, OnFailAction

```

```python
tox_guard = Guard().use(
    ToxicLanguage(
        threshold=0.5,
        validation_method="sentence",
        on_fail=OnFailAction.FIX,
        use_local=True, 
    )
)


rude = "You are completely useless and your stupid fuck app stole my money."

out = tox_guard.validate(rude)
```

```python
print(f"Validation Result: {out.validation_passed}")
print(f"Raw User Input: {out.raw_llm_output}")
print(f"Validated Output: {out.validated_output}")
```
